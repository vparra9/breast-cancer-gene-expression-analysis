```python
import os
from Bio import Entrez

# Always tell NCBI who you are
Entrez.email = os.getenv("NCBI_EMAIL")

# Search for the GSE accession number to get NCBI's internal ID
search_handle = Entrez.esearch(db="gds", term="GSE191230[Accession]")
search_results = Entrez.read(search_handle)
search_handle.close()

# Extract the internal ID assigned by NCBI
gds_id = search_results["IdList"][0]

# Fetch the metadata summary using that ID
summary_handle = Entrez.esummary(db="gds", id=gds_id)
metadata = Entrez.read(summary_handle)
summary_handle.close()

# Print the study title and summary description
print("Title:", metadata[0]["title"])
print("\nSummary:", metadata[0]["summary"])
```
