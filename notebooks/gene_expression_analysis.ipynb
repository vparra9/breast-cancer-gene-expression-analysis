```python
# Import the requests library to retrieve data from the GEO website
import requests

# GEO accession number for the sample we want to retrieve
sample_accession = "GSM5741915"

# NCBI GEO endpoint used to retrieve accession records
url = "https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi"

# Parameters requesting the full sample record in plain-text format
params = {
    "acc": sample_accession,
    "targ": "self",
    "view": "full",
    "form": "text"
}

# Send the request to NCBI GEO
response = requests.get(url, params=params)

# Display the returned GEO sample record
print(response.text)
```

```python


# Split the GEO record into individual lines
lines = response.text.splitlines()

# Find the sample description, which contains the dataset sample ID
for line in lines:
    if line.startswith("!Sample_description"):
        sample_id = line.split("=", 1)[1].strip()

    if line.startswith("!Sample_geo_accession"):
        geo_accession = line.split("=", 1)[1].strip()

    if line.startswith("!Sample_title"):
        sample_title = line.split("=", 1)[1].strip()

    if line.startswith("!Sample_source_name_ch1"):
        source = line.split("=", 1)[1].strip()

    if line.startswith("!Sample_characteristics_ch1"):
        characteristic = line.split("=", 1)[1].strip()

        if characteristic.startswith("tissue:"):
            tissue = characteristic.split(":", 1)[1].strip()

        if characteristic.startswith("cancer type:"):
            cancer_type = characteristic.split(":", 1)[1].strip()

    

print("Sample ID:", sample_id)
print("GEO Accession:", geo_accession)
print("Title:", sample_title)
print("Source:", source)
print("Tissue:", tissue)
print("Cancer Type:", cancer_type)
```

```python
# Function to retrieve and extract metadata for a GEO sample
def get_sample_metadata(sample_accession):

    # NCBI GEO endpoint used accession records
    url = "https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi"

    # Parameters requesting the full sample record in plain-text format
    params = {
        "acc": sample_accession,
        "targ": "self",
        "view": "full",
        "form": "text"
    }

    # Send the request to NCBI GEO
    response = requests.get(url, params=params)

    # Split the GEO record into individual lines
    lines = response.text.splitlines()

    # Extract relevant metadata fields from the GEO record
    for line in lines:
        if line.startswith("!Sample_description"):
            sample_id = line.split("=", 1)[1].strip()

        if line.startswith("!Sample_geo_accession"):
            geo_accession = line.split("=", 1)[1].strip()

        if line.startswith("!Sample_title"):
            title = line.split("=", 1)[1].strip()

        if line.startswith("!Sample_source_name_ch1"):
            source = line.split("=", 1)[1].strip()

        if line.startswith("!Sample_characteristics_ch1"):
            characteristic = line.split("=", 1)[1].strip()

            if characteristic.startswith("tissue:"):
                tissue = characteristic.split(":", 1)[1].strip()

            if characteristic.startswith("cancer type:"):
                cancer_type = characteristic.split(":", 1)[1].strip()

    # Return the extracted metadata as a dictionary
    return {
        "sample_id": sample_id,
        "geo_accession": geo_accession,
        "title": title,
        "source": source,
        "tissue": tissue,
        "cancer_type": cancer_type
    }
    


```

```python
# Retrieve metadata for all samples in the study
all_sample_metadata = []

for sample in metadata[0]["Samples"]:
    sample_accession = sample["Accession"]
    sample_info = get_sample_metadata(sample_accession)
    all_sample_metadata.append(sample_info)
```

```python
# Convert the sample metadata into a DataFrame
sample_metadata_df = pd.DataFrame(all_sample_metadata)

sample_metadata_df
```

```python
# Assign each sample to its tumor group based on the sample id
sample_metadata_df["group"] = sample_metadata_df["sample_id"].apply(
    lambda x: "Metastatic" if x.startswith("M") else "Primary"
)

sample_metadata_df
```
